```python
#Conditional Probability --> Measures the probabilty of an event when another event is
#already known to have occurred..

#Probability that a randomly selected student is passed?
#Probabilty that a student is Passed given that the student attended classes regularly

'''                          Passed.         Failed          Total
Regular Attendance  -->        48             12              60
Irregular.                     16             24              40
                            -------          -------
                               64             36             100
A -->Student Passed
B --> Student attending regulary
P(A)  = 64 / 100 = 0.64  #Probability of Passing
#Probability of passing given reg attendance.      P(A | B)  = 48 / 60  = 0.8

#P(A|B) = P(A and B)
        -----------------    #P(B) > 0
            P(B)


#P(A|B) --> probability of A given B
#P(A and B) --> Probability that both will occur.      --> 48 / 100. --> 0.48
#P(B) --> Probability of known condition             -->    60/100  --> 0.6
#P(A|B) = 0.48 / 0.6 --> 0.8

#Proability given regular attendance --> P(Passed | Regualr) --> 48/60 = 0.8
#probabilty regular attendance given pass --> P(Regular | Passed) --> 48/64 = 0.75
                 #P(A|B) != P(B|A)

P(AUB) = P(A) + P(B) - P(A and B). #Addition rule
P(A|B) = P(A and B)/P(B)
P(A and B) = P(A|B) * P(B).  #Mutliplication rule.  0.8 * 0.6 = 0.48
P(B and A) = P(A) * P(B|A) this can also happen
```

```python
#Given that a fair die is rolled an even number,find the probability that it is
#greater than 3
sample_space = {1,2,3,4,5,6}
even_count = 0
even_and_greater_than_3 = 0
for number in sample_space:
  #print(number)
  if number %2==0:
    #print(number)
    even_count +=1
    #print(even_count)
    if number > 3:
      even_and_greater_than_3+=1
      #print(even_and_greater_than_3)
probability = even_and_greater_than_3 / even_count
print(probability)
```

**Law of Total Probability = Student Results**

P(Passed) = P(Regular)*P(Passed|Regular) +
            P(Irregular)*P(Passed|Irregular)

```python
probability_regular = 0.6
probability_irregular = 0.4
passed_given_reg = 0.8
passed_given_irrg = 0.4
regular_and_passed = (probability_regular)*(passed_given_reg)
irregular_and_passed = (probability_irregular) *(passed_given_irrg)
probability_passed = (regular_and_passed + irregular_and_passed)
print(probability_passed)
print(f'P(Passed): {probability_passed}')
```

```python
failed_given_reg = 0.2   #12/60
failed_given_irreg = 0.6 #24/40
regular_and_failed = (probability_regular)*(failed_given_reg)
irregular_and_failed = (probability_irregular) *(failed_given_irreg)
probability_failed = (regular_and_failed + irregular_and_failed)
print(probability_failed)
print(f'P(Failed): {probability_failed}')
```
